```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from analysis.scripts.experiment_helpers import *
```

```python
exp3 = pd.read_json('analysis/data/experiments/exp3_run_20260606_211237.json')
runs = exp3["experiments"]

results = analyze_runs(runs)

print_grouped_results(results, by=('tau',))
```

```python
my_fig = plot_grouped_results(results, by=('tau', ), title="Measured delay per run by heartbeat interval Tau")
```

```python
oscillation_fig = plot_oscillation_traces(results, by=('tau', ), skip_initial=0, value_col='hardware_delay', figure_title="", y_label="error [ms]")
```

```python
if my_fig is not None:
    my_fig.savefig("ex3_hw_delay_plot.png", dpi=300, bbox_inches='tight')
if oscillation_fig is not None:
    oscillation_fig.savefig("ex3_oscillation_fig.png", dpi=300, bbox_inches='tight')
```
